```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
spy = pd.read_csv(
    "../data/spy_market_data.csv",
    index_col="Date",
    parse_dates=True
)

spy.head()
```

```python
spy["return_3d"] = spy["Adj Close"].pct_change(3)
spy["return_5d"] = spy["Adj Close"].pct_change(5)
spy["return_10d"] = spy["Adj Close"].pct_change(10)
spy["return_20d"] = spy["Adj Close"].pct_change(20)
```

```python
momentum_features = [
    "return_1d",
    "return_3d",
    "return_5d",
    "return_10d",
    "return_20d"
]

spy[momentum_features].tail(10)
```

```python
spy["sma_5"] = spy["Adj Close"].rolling(5).mean()
spy["sma_20"] = spy["Adj Close"].rolling(20).mean()
```

```python
spy["price_vs_sma5"] = spy["Adj Close"] / spy["sma_5"] - 1
spy["price_vs_sma20"] = spy["Adj Close"] / spy["sma_20"] - 1
```

```python
spy["sma5_vs_sma20"] = spy["sma_5"] / spy["sma_20"] - 1
```

```python
trend_features = [
    "price_vs_sma5",
    "price_vs_sma20",
    "sma5_vs_sma20"
]

spy[trend_features].tail(10)
```

```python
spy["volatility_5d"] = spy["return_1d"].rolling(5).std()
spy["volatility_20d"] = spy["return_1d"].rolling(20).std()
```

```python
spy["high_low_range"] = (
    spy["High"] - spy["Low"]
) / spy["Close"]
```

```python
spy["volatility_ratio"] = (
    spy["volatility_5d"] / spy["volatility_20d"]
)
```

```python
volatility_features = [
    "volatility_5d",
    "volatility_20d",
    "high_low_range"
]

spy[volatility_features].tail(10)
```

```python
spy[
    [
        "volatility_5d",
        "volatility_20d",
        "volatility_ratio",
        "high_low_range"
    ]
].tail(10)
```

```python
spy["overnight_gap"] = (
    spy["Open"] / spy["Close"].shift(1) - 1
)

spy["intraday_return"] = (
    spy["Close"] / spy["Open"] - 1
)
```

```python
spy["volume_sma20"] = spy["Volume"].rolling(20).mean()

spy["volume_ratio"] = (
    spy["Volume"] / spy["volume_sma20"]
)
```

```python
market_behavior_features = [
    "overnight_gap",
    "intraday_return",
    "volume_ratio"
]

spy[market_behavior_features].tail(10)
```

```python
volatility_features = [
    "volatility_5d",
    "volatility_20d",
    "volatility_ratio",
    "high_low_range"
]

feature_cols = (
    momentum_features
    + trend_features
    + volatility_features
    + market_behavior_features
)

print("Number of features:", len(feature_cols))
feature_cols
```

```python
modeling_df = spy[
    feature_cols + ["target_3d"]
].copy()

modeling_df.shape
```

```python
modeling_df.isna().sum()
```

```python
modeling_df.head(22)
```

```python
modeling_df_clean = modeling_df.dropna().copy()

print("Before:", modeling_df.shape)
print("After: ", modeling_df_clean.shape)

print("\nDate range:")
print(modeling_df_clean.index.min())
print(modeling_df_clean.index.max())
```

```python
#removing the last day

# Remove the current unfinished trading session
spy = spy.iloc[:-1].copy()

# Recompute the 3-day future return
spy["future_return_3d"] = (
    spy["Adj Close"].shift(-3) / spy["Adj Close"] - 1
)

# Recompute the target safely
spy["target_3d"] = np.nan

valid_target = spy["future_return_3d"].notna()

spy.loc[valid_target, "target_3d"] = (
    spy.loc[valid_target, "future_return_3d"] > 0.005
).astype(int)
```

```python
modeling_df = spy[
    feature_cols + ["target_3d"]
].copy()

modeling_df_clean = modeling_df.dropna().copy()

print("Shape:", modeling_df_clean.shape)
print("Start:", modeling_df_clean.index.min())
print("End:  ", modeling_df_clean.index.max())
```

```python
X = modeling_df_clean[feature_cols].copy()
y = modeling_df_clean["target_3d"].astype(int).copy()

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
target_counts = y.value_counts()
target_pct = y.value_counts(normalize=True)

print("Counts:")
print(target_counts)

print("\nPercentages:")
print(target_pct)
```

```python
modeling_df_clean.to_csv(
    "../data/spy_modeling_data.csv"
)

print("Saved:", modeling_df_clean.shape)
print("Features:", len(feature_cols))
print("Start:", modeling_df_clean.index.min())
print("End:", modeling_df_clean.index.max())
```
