## sklearn Logistic Regression — Summary

Our SPY Logistic Regression pipeline is:

\[
X_{raw}
\rightarrow
StandardScaler
\rightarrow
z=w^TX+b
\rightarrow
\sigma(z)
\rightarrow
P(y=1)
\]

The scaler standardizes each feature using training statistics:

\[
x_{scaled}=\frac{x-\mu_{train}}{\sigma_{train}}
\]

Logistic Regression computes:

\[
z=b+\sum_j w_jx_j
\]

and converts this raw score into a probability:

\[
P(y=1)=\frac{1}{1+e^{-z}}
\]

For one real SPY test observation, we manually calculated:

\[
z=-0.35797
\]

and:

\[
P(y=1)=0.411450689
\]

This exactly matched `sklearn.predict_proba()`.

Therefore, sklearn's prediction process is no longer a black box:
scaled features are multiplied by learned coefficients, summed with the intercept, and passed through the sigmoid function.

# Part 2 — Logistic Regression From Scratch

Now we rebuild Logistic Regression without `sklearn.LogisticRegression`.

We will implement:

1. Weight initialization
2. Linear score \(z = Xw + b\)
3. Sigmoid
4. Log loss
5. Gradients
6. Gradient descent
7. Training loop
8. Predictions
9. Comparison with sklearn

```python
# Scale using TRAIN statistics
X_train_s = scaler.transform(X_train)
X_test_s = scaler.transform(X_test)

# Convert targets to NumPy arrays
y_train_np = y_train.to_numpy()
y_test_np = y_test.to_numpy()

n_samples, n_features = X_train_s.shape

print("Samples:", n_samples)
print("Features:", n_features)
```

```python
def sigmoid(z):
    return 1 / (1+ np.exp(-z))

def log_loss(y, p):
    eps = 1e-15

    p = np.clip(p, eps, 1-eps)

    return -np.mean(
        y * np.log(p)
        + (1-y) * np.log(1-p)
    )
```

```python
weights = np.zeros(n_features)
bias = 0.0

learning_rate = 0.05
epochs = 2000

for epoch in range(epochs):
    #1. Linear score
    z = X_train_s @ weights + bias

    #2. Probability
    p = sigmoid(z)

    # 3. Gradients
    dw = (X_train_s.T @ (p-y_train_np)) / n_samples
    db = np.mean(p - y_train_np)

    # 4. Gradient descent update
    weights -= learning_rate * dw
    bias -= learning_rate * db

    # Show progress
    if epoch % 200 ==0:
        loss = log_loss(y_train_np, p)
        print(f"Epoch {epoch:4d} | Loss: {loss:.5f}")
```
